# Gerando dados por estação

```python
try: 
    estacoes = os.listdir(os.path.join(os.getcwd(), 'dados_chuva_goias'))
except NameError:
    estacoes = os.listdir(os.path.join(os.path.dirname(os.path.abspath(__file__)), 'dados_chuva_goias'))

df = {"codigo_estacao": [], "nome_estacao": [], "cidade": [], "latitude": [], "longitude": [], "altitude": [], "mês A.H.": [], "Qualidade (anos)": []}
for estacao in estacoes:
    print(f"Processando estação: {estacao}")
    # Acesso ao diretório atual e construção do caminho para os dados da estação
    try:
        diretorio_atual = os.path.dirname(os.path.abspath(__file__))
    except NameError:
        diretorio_atual = os.getcwd()
    fonte = os.path.join(diretorio_atual, 'dados_chuva_goias', estacao)

    # Leitura da estação e limpeza dos dados e resultados em termos de precipitação
    metadados, mensal, meses_secos, df_final, df_spi = ler_dados(fonte)
    print(f"Nome da estação: {metadados['nome']}")
    pasta_destino = os.path.join(diretorio_atual, 'resultados', f"{metadados['codigo_estacao']}_{metadados['nome']}")
    os.makedirs(pasta_destino, exist_ok=True)
    mensal.to_csv(os.path.join(pasta_destino, 'precipitacao_media_mensal.csv'), index=False)
    meses_secos.to_csv(os.path.join(pasta_destino, 'meses_secos.csv'), index=False)
    df_final.to_csv(os.path.join(pasta_destino, 'dados_limpos.csv'), index=False)
    df_spi.to_csv(os.path.join(pasta_destino, 'dados_spi.csv'), index=False)
    plot_precipitacao_media_mensal(pasta_destino, f"z_precipitacao_media_mensal_{metadados['codigo_estacao']}_{metadados['nome']}", mensal, inicio_ano_chuvoso=metadados['mes_inicio_ano_hidrologico'], lang='pt')
    plot_precipitacao_media_mensal(pasta_destino, f"z_precipitacao_media_mensal_{metadados['codigo_estacao']}_{metadados['nome']}", mensal, inicio_ano_chuvoso=metadados['mes_inicio_ano_hidrologico'], lang='en')

    # Precipitação máxima diária por ano hidrológico ou civil
    hmax1d = calcular_precipitacao_maxima_diaria(df_final)
    hmax1d.to_csv(os.path.join(pasta_destino, 'precipitacao_maxima_diaria.csv'), index=False)

    # Checagem da distribuição
    dist_df, parametros, nome_distribuicao = checar_dis_adequada(hmax1d)
    dist_df.to_csv(os.path.join(pasta_destino, 'teste_distribuicao.csv'), index=False)
    dist_obj = getattr(sc.stats, nome_distribuicao)
    dados_simulados_1000 = dist_obj.rvs(*parametros, size=1000, random_state=42)
    x_dados, y_dados = obter_cdf(hmax1d['precipitacao máxima diária do ano (mm)'].values)
    dist = getattr(sc.stats, nome_distribuicao)
    x_numerico = np.linspace(min(hmax1d['precipitacao máxima diária do ano (mm)'].values), max(hmax1d['precipitacao máxima diária do ano (mm)'].values), 1000)
    y_numerico = dist.cdf(x_numerico, *parametros)
    dados_h1max = {'real': hmax1d['precipitacao máxima diária do ano (mm)'].values, 'numerica': dados_simulados_1000}
    plot_pdf_precipitacao_maxima_diaria(pasta_destino, f"z_pdf_{metadados['codigo_estacao']}_{metadados['nome']}", dados_h1max, lang='pt')
    plot_pdf_precipitacao_maxima_diaria(pasta_destino, f"z_pdf_{metadados['codigo_estacao']}_{metadados['nome']}", dados_h1max, lang='en')
    dados_h1max_cdf = {'real': {'x': x_dados, 'y': y_dados}, 'numerica': {'x': x_numerico, 'y': y_numerico}}
    plot_cdf_precipitacao_maxima_diaria(pasta_destino, f"z_cdf_{metadados['codigo_estacao']}_{metadados['nome']}", dados_h1max_cdf, lang='pt')
    plot_cdf_precipitacao_maxima_diaria(pasta_destino, f"z_cdf_{metadados['codigo_estacao']}_{metadados['nome']}", dados_h1max_cdf, lang='en')

    # Tempo de retorno e máxima precipitação diária
    df_hmax = calcular_hmax(parametros, nome_distribuicao)
    df_hmax.to_csv(os.path.join(pasta_destino, 'hmax_tempo_retorno.csv'), index=False)

    # Matriz de chuva
    matriz_chuva = desagragacao_preciptacao_maxima_diaria_matriz_intensidade_chuva(df_hmax)
    matriz_chuva.to_csv(os.path.join(pasta_destino, 'matriz_chuva.csv'), index=True)
    plot_curvas_idf(pasta_destino, f"z_idf_{metadados['codigo_estacao']}_{metadados['nome']}", 'pt', matriz_chuva)
    plot_curvas_idf(pasta_destino, f"z_idf_{metadados['codigo_estacao']}_{metadados['nome']}", 'en', matriz_chuva)
    
    # Adicionando os metadados ao DataFrame
    df['codigo_estacao'].append(metadados['codigo_estacao'])
    df['nome_estacao'].append(metadados['nome'])
    df['cidade'].append(metadados['cidade'])

df.latex
```
